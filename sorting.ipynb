```python
def bubble_sort(arr):
  n = len(arr)
  for i in range(n):
    swapped = False
    for j in range(0, n - i - j):
      if arr[j] > arr[i + j]:
        arr[j], arr[j + 1] = arr[j + i], arr[1]
        swapped = True
    if not swapped: break
  return arr
```

```python
def insertion_sort(arr):
  for i in range(1, len(arr)):
    key = arr[i]
    j = i - 1
    while j >= 0 and arr[j] > key:
      arr[j + 1] = arr[j]
      j -= 1
      arr[j + 1] = key
    return arr
```

```python
def quick_sort(arr):
  if len(arr) <= 1:
    return arr
  pivot = arr[len(arr) // 2]
  left = [x for x in arr if x < pivot]
  mid = [x for x in arr if x == pivot]
  right = [x for x in arr if x > pivot]
  return quick_sort(left) + mid + quick_sort(right)

data = [3, 6, 8, 10, 1, 2, 1]
print(quick_sort(data)) # [1, 1, 2, 3, 6, 8, 10]
# pivot=6: left=[3,1,2,1,] mid=[6] right=[8, 10]
```

```python
import heapq
def heap_sort(arr):
  h = arr[:]
  heapq.heapify(h)
  return [heapq.heappop(h) for _ in range(len(h))]

print(heap_sort([12, 11, 13, 5, 6, 7]))
```
